```python
#Numpy miscellaneous functions
#arange() - used to generate 1Darray of numbers within specified interval
# with specified gap between consecutive values
print(np.arange(5,25,3))
print(np.arange(25))
print(np.arange(10,0,-2))
```

```python
#linspace- this function is used to generate 1Darray of numbers within 
# specified interval with specified gap betweem consecutive values 
#and of specified count
print(np.linspace(1,10,100))
```

```python
print(np.linspace(2,20,40))
```

```python
#numpy random submodule- used to generate array with random numbers
# rand()- generates any dimensional array of random floating 
# point number between half open interval of(0.1)
print(np.random.rand(5))
print(np.random.rand(5,3))
print(np.random.rand(3,2,4))
```

```python
#Random - generates any dimensional array of random floating point number between half open 
#interval of (0,1] requires the shape to be given as a tuple for 2d or more dimensional arrays
print(np.random.random(5))
print("")
print(np.random.random((3,2)))
print("")
print(np.random.random(((2,3,4))))
```

```python
#uniform() -generate random floating number array of specified shape within 
#the specified interval, which follow uniform distribution
print(np.random.uniform(2,10,size=(10)))
```

```python
#Randint()- Generates random integer number array of specified shape wiyhin the specified interval
print(np.random.randint(2,20,size=(10)))
```

```python
#Seed()
np.random.seed(10)
print(np.random.randint(2,20,size=(10)))
print("")
np.random.rand()
```
