### Data cleaning


```python
df.isnull().sum()
```

```python
df.dropna() # na drops all rows with null values
```

```python
df.dropna(how="any") # any row that had any null value
```

```python
df.dropna(how="all") # if all values in a row are null otherwise will not be dropped
```

```python
# Filling null values
df["Age"].fillna(df["Age"].mean())
```

```python
df["Salary"].fillna(df["Salary"].median())
```

```python
# forward fill- top to bottom
df["Age"].bfill()
```

```python
df["Name"]=df["Name"].replace("Charlie", "Rose")
df
```

```python
# Dealing with duplicate values
df_dup=df[df.duplicated(keep="first")] # first occurrence is kept and rest are marked as duplicates
df_dup
```

```python
#Invalid values
df["Promoted salary"]=df["Promoted salary"].apply(lambda x: x/10 if x>65000 else x)
df
```

```python
#apply and lambda
def multiplying_age(x):
    return x*2
df["Age"]=df["Age"].apply(multiplying_age)
df
```

```python
df["Age"]=df['Age'].apply(lambda x:x/2)
df
```

```python
# Joins and merges
# left, right, inner, outer joins
department_info={
    "Department":["HR","IT","Finance"],
    "Location":["New York","San Francisco","Chicago"],
    "Manager":["John","Alice","Bob"]
}
df2=pd.DataFrame(department_info)
df2
```

```python
# Concat
pd.concat([df,df2])
```

```python
pd.merge(df, df2, on="Dept")
```
