```python
# using dictionaries
fruit_protein = {
    "guava": 2.0,
    "blackberries": 2.6,
    "oranges": 0.9,
    "banana": 1.1,
    "apples": 0.3,
    "kiwi": 1.1,
    "pomegranate": 1.7,
    "mango": 1.7,
    "cherries": 1.0
}
```

```python
s2 = pd.Series(fruit_protein)
s2
```

```python
#Conditional seection
s2[s2>1]
```

```python
#logical operators : and, or,, not
s2[(s2>0.5) & (s2<2)]
```

```python
#modifying series
s2["mango"]=2.8
s2
```

### DataFrame

```python
data={
    "Name":["Alice","Bob","Charlie","David","Eve","Alice"],
    "Age":[25,30,35,np.nan,29,25],
    "Department":["HR","IT","Finance","IT","HR","HR"],
    "Salary":[50000,60000,70000,62000,np.nan,50000]
}
```

```python
data
```

```python
df=pd.DataFrame(data)
df
```

```python
df.head(2)
```

```python
df.tail(3)
```

```python
#loc & iloc
df.iloc[1:3]
```

```python

df.loc[1:3,["Age","Department"]]
```

```python
# accesing individual column
df[["Age","Department"]]
```

```python
# Dropping column
df.drop("Age",axis=1) # axis = 0 if we're dropping particular row
```

```python
df.shape
```

```python
df.dtypes
```

```python
df.info()
```

```python
# Describe - gives ststistical information
df.describe()
```

```python
# Broadcasting
df["Salary"]=df["Salary"]+5000
df["Salary"]
```

```python
# Renaming columns
df.rename(columns={"Department": "Dept"},inplace=True)
df
```
