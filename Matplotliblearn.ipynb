```python
# Pyplot API
# Univariate - numerical
data={
    "salary":[25000,30000,37000,28000,48000,27000,26000,31000,49000,23000,38000,57000,55000,52000,50000],
    }
df=pd.DataFrame(data)
```

```python
df.head()
```

```python
#line plot
plt.plot(df["salary"],color="blue", marker="o",linestyle=":",linewidth=2,markersize=8)
plt.grid()
plt.show() 
```

```python
# Histogram
plt.hist(df["salary"],bins=5,color="green",edgecolor="black")
plt.show()
```

```python
plt.boxplot(df["salary"])
plt.show()
```

```python
# Univariate - categorical
df["department"] = ["IT","HR","IT","Finance","Finance","IT","HR","Finance","IT","HR","Finance","IT","HR","Finance","IT"]
```

```python
df.head()
```

```python
df.info()
```

```python
#pie chart
count=df.department.value_counts()
count
```

```python
plt.pie(count,labels=count.index,autopct="%1.1f",explode=[0.1,0.1,0.1],shadow=True)
plt.show()
```

```python
plt.bar(count.index,count,color="green",edgecolor="black")
plt.show()
```

```python
#Bivariate - numerical
df["Age"] = [25,30,35,28,40,27,26,31,45,23,38,57,55,52,50]
```

```python
df.head()
```

```python
# Scatter plot  
plt.scatter(df["Age"],df["salary"],color="blue")
plt.show()
```

```python
#Line plot
sort_age=df.sort_values("Age")
plt.plot(sort_age["Age"],df["salary"],color="blue", marker="o",linestyle=":",linewidth=2,markersize=8)
plt.show()
```

```python
df["experience"] = [2,5,10,3,15,1,2,6,20,1,12,25,22,18,15]
df.head()
```

```python
#Bubble plot
plt.scatter(df["Age"],df["salary"],s=df["experience"]*10,color="blue")
plt.title("Age vs salary vs experience")
plt.xlabel("Age")
plt.ylabel("salary")
plt.grid()
plt.show()
```

```python
# 2 numerical and 1 categorical column
plt.scatter(df["Age"],df["salary"],s=df["experience"]*10,c=df["department"].map({"IT":"blue","HR":"green","Finance":"red"}))
plt.xlabel("Age")
plt.ylabel("salary")
plt.title("Age vs salary vs department")
plt.legend()
plt.show()
```

```python
# 2 numerical & 1 categorical column
color={"HR":"green","IT":"blue","Finance":"red"}
for department, color in color.items():
    df_dept = df.loc[df["department"] == department]
    plt.scatter(df_dept["Age"],df_dept["salary"],s=df_dept["experience"]*10,color=color,label=department)
plt.legend() 
```
