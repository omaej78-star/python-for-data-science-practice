```python
import numpy as np
import pandas as pd
```

```python
wardrobe_arr=np.array([3,4,2,3,2,1,2,4])
print(wardrobe_arr)
```

```python
wardrobe_sr=pd.Series(wardrobe_arr)
print(wardrobe_sr)
```

```python
wardrobe_sr_with_index=pd.Series(wardrobe_arr,index= ['Trousers','Shirts','Socks',
                                                     'Shoes','Belt','Tie',
                                                     'Handkerchief','Wallet'])

print(wardrobe_sr_with_index)
```

```python
#A series can either be created as the series class object or it could be generated
# through other  methods
# The generic syntax to create series is:
# pd.series(data,<list of Indices>)
```

```python
#Creating series from a list
data=[1,2,3,4,5,6]
sr1=pd.Series(data)

##Creating series from a list with specified index
sr1_with_index=pd.Series(data, index=['a', 'b','c','d','e','f'])
print (sr1)
print('*'*50)
print(sr1_with_index)
```

```python
#Creating series from an array
data=np.array([1,2,3,4,5,6])
sr1=pd.Series(data)

##Creating series from an array with specified index
sr1_with_index=pd.Series(data, index=['a', 'b','c','d','e','f'])
print (sr1)
print('*'*50)
print(sr1_with_index)
```

```python
#Create a series from a dictionary
dict1={'a': 1,'b': 2,'c': 3,'d': 4,'y': 5,'z': 6}
sr_from_dict=pd.Series(dict1)
print(sr_from_dict)
```

```python
#Create a series from a dictionary with index
dict1={'a': 7,'b': 2,'c': 13,'d': 16,'y': 5,'z': 6}
sr_from_dict_with_index=pd.Series(dict1,index=['a', 'b','g','d','e','f'])

print(sr_from_dict_with_index)
print(sr_from_dict)
```

```python
wardrobe_sr_with_index=pd.Series(wardrobe_arr,index= ['Trousers','Shirts','Socks',
                                                     'Shoes','Belt','Tie',
                                                     'Handkerchief','Wallet'])

print(wardrobe_sr_with_index)
```

```python
print(wardrobe_sr_with_index.loc['Shirts'])
# loc stands for location and strictily need user to define labels
print(wardrobe_sr_with_index.iloc[1])
# iloc stands for integer location and strictily need integer index/labels
```

```python
# Accessing multiple values
print(wardrobe_sr_with_index.loc[['Shirts','Belt','Wallet']])
print()
print(wardrobe_sr_with_index.iloc[[1,5,4]])
print()
print(wardrobe_sr_with_index.iloc[1:5])
```

```python
#Basic panda functios -Series
print(wardrobe_sr_with_index)
```

```python
print(wardrobe_sr_with_index.ndim) # number of dimesion = 1
print(wardrobe_sr_with_index.shape) # number of elements in each dimension as tuple
print(wardrobe_sr_with_index.size) # total number of elements
print(wardrobe_sr_with_index.dtype) # data type of series
print(wardrobe_sr_with_index.index) # gives the array of indices
print(wardrobe_sr_with_index.values) # gives data as 1d array
```

```python
# Head() - Returns top n data points : dafault 5
print(wardrobe_sr_with_index.head())
print()
print(wardrobe_sr_with_index.head(2))
```

```python
# Tail() - Returns last n data points : dafault 5
print(wardrobe_sr_with_index.tail())
print()
print(wardrobe_sr_with_index.tail(2))
```

```python
# info() - gives the general summary
print(wardrobe_sr_with_index.info())
```

```python
# describe() - gives the statistical summary
print(wardrobe_sr_with_index.describe())
```

```python
# unique() - giveas the array of unique data
print(wardrobe_sr_with_index.unique())
```

```python
# nunique() - gives the count of unique data
print(wardrobe_sr_with_index.nunique())
```

```python
# value_counts() - gives the frequency of unique data
print(wardrobe_sr_with_index.value_counts())
```

```python
# sample() - shows n random data points : default is 1
print(wardrobe_sr_with_index.sample())
print()
print(wardrobe_sr_with_index.sample(3))
```

```python
# isnull()/isna() - returns true for every missing data else false
print(sr_from_dict_with_index.isnull())
```

```python
# Find the count of missng values - ones that are true
print(sr_from_dict_with_index.isnull().sum())
```

```python
# Treat the missing values - Fill missing values with appropriate data - use fillna()
sr_from_dict_with_index=sr_from_dict_with_index.fillna('Missing data')
print(sr_from_dict_with_index)
```

```python
#DataFrame - A 2d tabular data structure. It is the primary data structure thet is used to handle and manipulate data
#A dataFrame can be created as an object of Pandas class 'DataFrame or by reading external data sources like csv, excel,html...'

data_list = [['Alice',25,50000],['Bob',30,60000],['Charlie',22,45000]]
clmns=['Name','Age','Salary']
idx=['emp1','Emp2','Emp3']
  
df1= pd.DataFrame(data_list, columns=clmns,index=idx)
df1
```

```python
#creating DataFrame using array
data_list = np.array([['Alice',25,50000],['Bob',30,60000],['Charlie',22,45000]])
clmns=['Name','Age','Salary']
idx=['emp1','Emp2','Emp3']
  
df2= pd.DataFrame(data_list, columns=clmns,index=idx)
df2
```
